Clasificación de calificaciones

Escribe un programa en Python que solicite al usuario una calificación numérica entre 0 y 10 y la clasifique de la siguiente forma:

Rango	Clasificación
Menor que 6	NA
De 6 a menor que 7.6	S
De 7.6 a menor que 8.6	B
De 8.6 a 10	MB
El programa debe:

Pedir la calificación usando input(). Convertir el valor ingresado a tipo float. Usar if, elif y else. Imprimir la clasificación correspondiente. Si la calificación es menor que 0 o mayor que 10, imprimir: "Calificación no válida"

```python
cal=float(input("Ingresa un número: "))
if cal<0 or cal>10: 
    print("Calificacion no válida")
elif cal>=8.6 and cal<=10:
    print ("MB")
elif cal >=7.6 and cal<=8.6:
    print("B")
elif cal>=6 and cal<=7.6:
    print ("S")
else:
    print("NA")
```
