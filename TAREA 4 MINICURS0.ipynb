```python
# IMPORTAMOS PAQUETERIAS
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# ANEXAMOS LA RUTA DE LOS ARCHIVOS
homicidios_2017='https://raw.githubusercontent.com/jamc88/min-curso/refs/heads/main/Datos/homicidios_2017.csv'
homicidios_2018='https://raw.githubusercontent.com/jamc88/min-curso/refs/heads/main/Datos/homicidios_2018.csv'
homicidios_2019='https://raw.githubusercontent.com/jamc88/min-curso/refs/heads/main/Datos/homicidios_2019.csv'
homicidios_2020='https://raw.githubusercontent.com/jamc88/min-curso/refs/heads/main/Datos/homicidios_2020.csv'
homicidios_2021='https://raw.githubusercontent.com/jamc88/min-curso/refs/heads/main/Datos/homicidios_2021.csv'
homicidios_2022='https://raw.githubusercontent.com/jamc88/min-curso/refs/heads/main/Datos/homicidios_2022.csv'
homicidios_2023='https://raw.githubusercontent.com/jamc88/min-curso/refs/heads/main/Datos/homicidios_2023.csv'
```

```python
# LEEMOS LOS DATAFRAME
df2017=pd.read_csv(homicidios_2017)
df2018=pd.read_csv(homicidios_2018)
df2019=pd.read_csv(homicidios_2019)
df2020=pd.read_csv(homicidios_2020)
df2021=pd.read_csv(homicidios_2021)
df2022=pd.read_csv(homicidios_2022)
df2023=pd.read_csv(homicidios_2023)
```

```python
# AGREGAMOS  UNA NUEVA COLUMNA
df2017['Año']=2017
df2018['Año']=2018
df2019['Año']=2019
df2020['Año']=2020
df2021['Año']=2021
df2022['Año']=2022
df2023['Año']=2023
```

```python
# UNIMOS TOAS LAS BASES DE DATOS EN UN SOLO DATAFRAME
df_homicidios=pd.concat([df2017, df2018, df2019, df2020, df2021, df2021, df2022, df2023], ignore_index=True)
df_homicidios
```

```python
# CREAMOS UNA COLUMNA TOTAL 
df_homicidios['Total']= df_homicidios['Mujeres'] + df_homicidios['Hombres'] + df_homicidios['No_esp']
```

```python
# OBSERVAMOS LA INFORMACIÓN CONTENIDA
df_homicidios.head()
```

```python
# ¿Cuántos registros tiene la base completa?
len(df_homicidios)
```

```python
# ¿Cuántos homicidios se registraron en total entre 2017 y 2023?
df_homicidios['Total'].sum()
```

```python
# ¿Qué año tuvo el mayor número total de homicidios?
total_año=df_homicidios.groupby('Año')['Total'].sum()
total_año.idxmax()
# USAMOS IDXMAX PARA SABER DÓNDE ESTABA EL MÁXIMO DEL ÍNDICE AÑO
```

```python
# ¿Qué año tuvo el menor número total de homicidios?
total_año.idxmin()
```
