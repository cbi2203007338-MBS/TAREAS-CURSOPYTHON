Tarea 1
Simula el lanzamiento de dos dados justos durante 1000 repeticiones.

Realiza lo siguiente:

Usa la paquetería random.
Guarda en una lista la suma de los dos dados en cada lanzamiento.
Cuenta cuántas veces aparece cada suma posible, desde 2 hasta 12.
Calcula la frecuencia relativa de obtener una suma igual a 7.
Grafica un histograma de las sumas obtenidas.
¿Por qué la suma 7 aparece con más frecuencia que la suma 2 o la suma 12?
 

```python
import random
```

```python
dado_1= [1, 2, 3, 4, 5, 6]
dado_2= [1, 2, 3, 4, 5, 6]
lista=[]
for i in range(1000):
    valor_1= random.choice(dado_1)
    valor_2= random.choice(dado_2)
    suma= valor_1 + valor_2
    lista.append(suma)
print(lista)
```

```python
for i in range(2,12):
    print(i, "aparece", lista.count(i),"veces")
```

```python
frecuencia_relativa=lista.count(7)/len(lista)
frecuencia_relativa
```

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(lista, bins=12, edgecolor="black")
plt.xlabel("Cara de los dados")
plt.ylabel("Suma de los dados")
plt.title("Suma del lanzamiento de dos dados,1000 veces")
plt.show()
```

¿Por qué la suma 7 aparece con más frecuencia que la suma 2 o la suma 12?

porque la suma de 7, tiene más combinaciones posibles, porque la lanzarlo 1000 veces la suma aparecio 156 veces mientras que la suma 2 y 12 aparecieron mucho menos, y esto ocurre porque el 7 tiene de 6 maneras posibles.

# Ejercicio 2


Usa NumPy para generar 300 puntos en el intervalo [0,4] y evalua la función

 

Realiza lo siguiente:

Crea el arreglo de puntos usando np.linspace().
Evalúa la función usando operaciones vectorizadas de NumPy.
Calcula: valor mínimo de la función, valor máximo de la función, media y desviación estándar.
Grafica la función usando plt.plot().
Agrega título, etiquetas en los ejes y cuadrícula.

```python
import numpy as np
```

```python
puntos=np.linspace(0,4*np.pi,300)
puntos
```

```python
funcion= np.sin(puntos)+ 1/2*np.cos(3*puntos)
funcion
```

```python
minimo=np.min(funcion)
maximo=np.max(funcion)
media=np.mean(funcion)
desviacion=np.std(funcion)

print('Valor minimo:', minimo)
print('Valor máximo:', maximo)
print('Media', media)
print('La desviacion estandar es:', desviacion)
```

```python
plt.figure(figsize=(8,5))
plt.plot(puntos,funcion)
plt.title('f(x)=sin(x)+1/2cos(3x)')
plt.xlabel('Puntos')
plt.ylabel('Funcion')
plt.grid()
plt.show()
```
